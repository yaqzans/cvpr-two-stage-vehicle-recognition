```python
import glob
from collections import Counter

root = None
for p in glob.glob("/kaggle/input/**/rsud20k-bangladesh-road-scene-understanding", recursive=True):
    root = p
    break
print("dataset root:", root)
assert root, "dataset not attached"

print("\ntop-level entries:")
for p in sorted(glob.glob(f"{root}/*")):
    print(" ", p)

print("\nall subfolders under root (depth <=4), with file counts:")
counts = Counter()
for p in glob.glob(f"{root}/**/*", recursive=True):
    parts = p[len(root):].strip("/\\").split("/")
    if len(parts) <= 1:
        continue
counter = Counter()
import os
for dirpath, dirnames, filenames in os.walk(root):
    rel = dirpath[len(root):]
    depth = rel.count(os.sep)
    if depth <= 3 and filenames:
        counter[rel] = len(filenames)
for k in sorted(counter):
    print(f"  {k!r}: {counter[k]} files")

print("\nsample files under root (first 30, recursive):")
n = 0
for dirpath, dirnames, filenames in os.walk(root):
    for f in filenames:
        print(" ", os.path.join(dirpath, f))
        n += 1
        if n >= 30:
            break
    if n >= 30:
        break

print("\nsearching for any .txt label-like files anywhere under root:")
n = 0
for p in glob.glob(f"{root}/**/*.txt", recursive=True):
    print(" ", p)
    n += 1
    if n >= 15:
        break
print("total .txt files found (capped at scan):", n)
```
